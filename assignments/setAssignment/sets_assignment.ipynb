```python
# set of first 10 +ve integers
myset={nums for nums in range(0,11)}
print(myset)
# add and delete
myset.add(11)
myset.remove(1)
print(myset)
myset={even_num for even_num in myset if even_num%2==0}
print(myset)
```

```python
#different operations
set1={1,2,3,4,5}
set2={0,2,4,6,8}
union_set=set1.union(set2)
print(union_set)
intersect_set=set1.intersection(set2)
print(intersect_set)
diffset1=set1.difference(set2)
diffset2=set2.difference(set1)
symm_diff=set1.symmetric_difference(set2)
print(diffset1)
print(diffset2)
print(symm_diff)
```
